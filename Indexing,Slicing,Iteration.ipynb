```python
import numpy as np
```

```python
arr1=np.arange(24).reshape(6,4)
arr1
```

```python
arr2=np.arange(1,6)
arr2
```

```python
arr2[0]
```

```python
#row indexing
arr1[0]
```

```python
# row slicing
arr1[1:3]
```

```python
#column slicing
arr1[:,0:3]
```

```python
# Indexing
arr1[0][3]
```

```python
arr1[4:6,2:4]
```

```python
for i in arr1:
  print(i)
```

```python
for i in np.nditer(arr1):
  print(i)
```
