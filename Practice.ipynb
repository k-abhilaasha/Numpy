Level 7
Data Analyst-style problem

```python
attendance = np.array([
    [85, 90, 78, 92],
    [70, 65, 80, 75],
    [95, 88, 91, 94],
    [60, 72, 68, 70]
])
```

```python
# average of each student
avg=[]
for i in attendance:
  print(np.average(i))
  avg.append(np.average(i))

```

```python
attendance.mean(axis=1)  # each student
```

```python
attendance.mean(axis=0)  # each subject
```

```python
# average of each subject
for i in attendance.T:
  print(np.average(i))
```

```python
np.argmax(avg)
```

```python
# Find which student has the highest average.
np.argmax(attendance.mean(axis=1))
```

```python
# Find which subject has the lowest average.
np.argmin(attendance.mean(axis=0))
```

```python
# Find all attendance values below 75.
attendance[attendance<75]
```

```python
# Count how many attendance values are below 75.
np.sum(attendance<75)
```

```python
import numpy as np

sales = np.array([
    [1200, 1500, 1100, 1800],
    [900, 1300, 1250, 1600],
    [2000, 2100, 1900, 2200]
])

# 1. Total sales of each salesperson
total_person = sales.sum(axis=1)
print("Total sales of each salesperson:", total_person)

# 2. Average sales of each salesperson
avg_person = sales.mean(axis=1)
print("Average sales of each salesperson:", avg_person)

# 3. Total sales for each month
total_month = sales.sum(axis=0)
print("Total sales for each month:", total_month)

# 4. Best salesperson
best_person = np.argmax(total_person)
print("Best salesperson:", best_person)

# 5. Best month
best_month = np.argmax(total_month)
print("Best month:", best_month)

# 6. All sales above 1500
above_1500 = sales[sales > 1500]
print("Sales above 1500:", above_1500)

# 7. Number of sales above 1500
count = np.sum(sales > 1500)
print("Number of sales above 1500:", count)

# 8. Overall average sales
overall_avg = sales.mean()
print("Overall average sales:", overall_avg)
```
