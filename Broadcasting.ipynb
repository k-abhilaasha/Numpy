```python
import numpy as np
```

```python
# if both arrays are same shape then boradcasting is not needed
a1=np.arange(6).reshape(3,2)
a2=np.arange(6,12).reshape(3,2)
```

```python
a1+a2
```

```python
# if x=1,y=m then broadcating takes place
a3=np.arange(3).reshape(1,3)
a4=np.arange(8,14).reshape(2,3)
a3+a4
```

```python
# if x=m,y=1 then broadcating takes place
a5=np.arange(3).reshape(3,1)
a6=np.arange(8,14).reshape(3,2)
a5+a6
```

```python
# if x=1,y=1 it will takes place
a7=np.arange(1).reshape(1,1)
a8=np.arange(8,14).reshape(3,2)
a7+a8
```

```python
# if x=1 and y!=n then also operation does not takes place
a9=np.arange(3).reshape(1,3)
a10=np.arange(16).reshape(4,4)
a9+a10
```

```python
# if they are different dimensions
a11=np.arange(4)
a12=np.arange(20).reshape(5,4)
a11+a12
```
